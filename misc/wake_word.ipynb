```python
from scipy import signal
import matplotlib.pyplot as plt
from assistant.utils import load_greeting
import numpy as np

window_size = 20
step_size = 10

greeting = load_greeting(voice="ash", idx=1)
nperseg = round(window_size * greeting.sample_rate / 1e3)
noverlap = round(step_size * greeting.sample_rate / 1e3)
freqs, times, spec = signal.spectrogram(
    greeting.samples,
    fs=greeting.sample_rate,
    window="hann",
    nperseg=nperseg,
    noverlap=noverlap,
)
log_spec = np.log(spec + 1e-10)

plt.pcolormesh(times, freqs, log_spec)
plt.ylabel("Frequency [Hz]")
plt.xlabel("Time [sec]")
plt.show()
```
